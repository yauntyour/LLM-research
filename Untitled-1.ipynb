```python
import torch
import torch.nn as nn
import torch.nn.functional as F


class SwiGLU(nn.Module):
    def __init__(self, in_dim, out_dim=None):
        super().__init__()
        out_dim = out_dim or in_dim
        self.fc = nn.Linear(in_dim, 2 * out_dim)

    def forward(self, x):
        x = self.fc(x)
        gate, value = x.chunk(2, dim=-1)
        return F.silu(gate) * value


class LinearAttention(nn.Module):
    def __init__(self, D: int):
        super().__init__()
        self.Q = nn.Sequential(nn.Linear(D, D), nn.Sigmoid())
        self.K = nn.Sequential(nn.Linear(D, D), nn.Sigmoid())
        self.V = nn.Linear(D, D, bias=False)

        self.W = nn.Sequential(nn.Linear(D, D), nn.Sigmoid())

        self.ffn = SwiGLU(D)
        self.norm = nn.RMSNorm(D)

    def forward(self, x):
        Q = self.Q(x)
        K = self.K(x)
        V = self.V(x)

        beta = K.transpose(-2, -1) @ V
        att = Q @ beta / x.shape[-2]
        W = self.W(att)
        x = self.norm(self.ffn(att) * W + x)
        return x


class Decoder(nn.Module):
    def __init__(self, D: int, n_layers: int):
        super().__init__()
        self.blocks = nn.ModuleList([LinearAttention(D) for _ in range(n_layers)])

    def forward(self, x):
        for block in self.blocks:
            x = block(x)
        return x
```
